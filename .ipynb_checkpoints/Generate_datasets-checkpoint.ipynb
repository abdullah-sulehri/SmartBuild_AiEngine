```python
import logging
from datetime import datetime, timedelta
import numpy as np
import pandas as pd
import os
from pathlib import Path

# ==========================================
# 1. CONFIGURATION (Directly inside the notebook)
# ==========================================
RANDOM_SEED = 42
DATA_SOURCE_TAG = "synthetic-illustrative"
TREND_WEEKS = 191  # <-- UPDATED: 191 weeks brings the data up to Sept 2026
TREND_START_DATE = "2023-01-02"
MATERIALS = ["Cement", "Steel", "Bricks", "Sand", "Aggregate"]

MATERIAL_BASE_PRICE = {"Cement": 1300.0, "Steel": 250000.0, "Bricks": 14000.0, "Sand": 40000.0, "Aggregate": 45000.0}
MATERIAL_UNIT = {"Cement": "PKR/50kg bag", "Steel": "PKR/ton", "Bricks": "PKR/1000 bricks", "Sand": "PKR/100 cft", "Aggregate": "PKR/100 cft"}
MATERIAL_ANNUAL_INFLATION = {"Cement": 0.12, "Steel": 0.18, "Bricks": 0.10, "Sand": 0.08, "Aggregate": 0.08}
MATERIAL_MONSOON_DIP = {"Cement": 0.02, "Steel": 0.015, "Bricks": 0.03, "Sand": 0.06, "Aggregate": 0.05}
MONSOON_MONTHS = (7, 8, 9)
MATERIAL_NOISE_STD_FRAC = {"Cement": 0.01, "Steel": 0.015, "Bricks": 0.012, "Sand": 0.02, "Aggregate": 0.02}

NUM_ORDERS = 5000
NUM_SUPPLIERS_FOR_ORDERS = 400
TRUCK_TYPES = ["mini", "6-wheeler", "10-wheeler"]
TRUCK_BASE_FEE = {"mini": 3000.0, "6-wheeler": 6000.0, "10-wheeler": 9500.0}
TRUCK_PER_KM_RATE = {"mini": 55.0, "6-wheeler": 85.0, "10-wheeler": 120.0}
URGENCY_LEVELS = ["standard", "express"]
URGENCY_MARKUP_FRAC = {"standard": 0.0, "express": 0.12}

CITY_DISTANCES_FROM_LAHORE_KM = {
    "Lahore": 15.0, "Sheikhupura": 45.0, "Kasur": 55.0, "Gujranwala": 80.0,
    "Sialkot": 125.0, "Faisalabad": 130.0, "Sargodha": 190.0, "Gujrat": 155.0,
    "Multan": 340.0, "Rawalpindi": 275.0,
}

QUANTITY_RANGE_BY_MATERIAL = {
    "Cement": (50, 2000), "Steel": (1, 40), "Bricks": (5, 200),
    "Sand": (2, 60), "Aggregate": (2, 60),
}

NUM_SUPPLIERS = 1000
SUPPLIER_QUALITY_MEAN = 0.72
SUPPLIER_QUALITY_STD = 0.16

# Setup Directories
os.makedirs("data", exist_ok=True)

# ==========================================
# 2. GENERATION FUNCTIONS
# ==========================================
def generate_price_trends(rng):
    start_date = datetime.strptime(TREND_START_DATE, "%Y-%m-%d")
    rows = []
    for material in MATERIALS:
        base_price = MATERIAL_BASE_PRICE[material]
        annual_inflation = MATERIAL_ANNUAL_INFLATION[material]
        monsoon_dip = MATERIAL_MONSOON_DIP[material]
        noise_std_frac = MATERIAL_NOISE_STD_FRAC[material]
        unit = MATERIAL_UNIT[material]

        for week in range(TREND_WEEKS):
            date = start_date + timedelta(weeks=week)
            years_elapsed = week / 52.0
            trend_price = base_price * ((1.0 + annual_inflation) ** years_elapsed)
            if date.month in MONSOON_MONTHS:
                trend_price *= (1.0 - monsoon_dip)
            noise = rng.normal(loc=0.0, scale=noise_std_frac * trend_price)
            price = max(trend_price + noise, 1.0)

            rows.append({
                "date": date.strftime("%Y-%m-%d"), "material": material,
                "price": round(price, 2), "unit": unit, "data_source": DATA_SOURCE_TAG
            })
    return pd.DataFrame(rows)

def _lookup_price(price_df, material, order_date):
    subset = price_df[price_df["material"] == material]
    subset_dates = pd.to_datetime(subset["date"])
    idx = (subset_dates - order_date).abs().idxmin()
    return float(subset.loc[idx, "price"])

def generate_landed_cost_orders(rng, price_df):
    start_date = datetime.strptime(TREND_START_DATE, "%Y-%m-%d")
    total_span_days = (start_date + timedelta(weeks=TREND_WEEKS) - start_date).days
    cities = list(CITY_DISTANCES_FROM_LAHORE_KM.keys())
    supplier_ids = [f"SUP-{i:04d}" for i in range(1, NUM_SUPPLIERS_FOR_ORDERS + 1)]

    rows = []
    for order_num in range(1, NUM_ORDERS + 1):
        material = rng.choice(MATERIALS)
        qty_min, qty_max = QUANTITY_RANGE_BY_MATERIAL[material]
        quantity = round(float(rng.uniform(qty_min, qty_max)), 1)
        
        day_offset = int(rng.integers(0, total_span_days))
        order_dt = start_date + timedelta(days=day_offset, hours=int(rng.integers(8, 18)), minutes=int(rng.integers(0, 60)))
        
        city = rng.choice(cities)
        distance_km = CITY_DISTANCES_FROM_LAHORE_KM[city]
        distance_km = max(distance_km + rng.normal(0, distance_km * 0.05), 1.0)

        qty_frac = (quantity - qty_min) / max(qty_max - qty_min, 1e-9)
        truck_probs = np.array([0.6, 0.3, 0.1]) if qty_frac < 0.33 else np.array([0.25, 0.5, 0.25]) if qty_frac < 0.66 else np.array([0.05, 0.35, 0.6])
        truck_type = rng.choice(TRUCK_TYPES, p=truck_probs)
        urgency = rng.choice(URGENCY_LEVELS, p=[0.8, 0.2])
        
        base_price = _lookup_price(price_df, material, order_dt)
        material_cost = base_price * quantity
        freight_cost = TRUCK_BASE_FEE[truck_type] + TRUCK_PER_KM_RATE[truck_type] * distance_km
        urgency_markup = material_cost * URGENCY_MARKUP_FRAC[urgency]
        
        rows.append({
            "order_id": f"ORD-{order_num:06d}", "order_timestamp": order_dt.strftime("%Y-%m-%d %H:%M:%S"),
            "material": material, "quantity": quantity, "unit": MATERIAL_UNIT[material],
            "destination_city": city, "distance_km": round(distance_km, 1), "truck_type": truck_type,
            "urgency": urgency, "supplier_id": rng.choice(supplier_ids),
            "base_material_price": round(base_price, 2), "material_cost": round(material_cost, 2),
            "freight_cost": round(freight_cost, 2), "urgency_markup": round(urgency_markup, 2),
            "landed_cost": round(material_cost + freight_cost + urgency_markup, 2), "data_source": DATA_SOURCE_TAG
        })
    return pd.DataFrame(rows)

def generate_supplier_trust(rng):
    n = NUM_SUPPLIERS
    quality = np.clip(rng.normal(SUPPLIER_QUALITY_MEAN, SUPPLIER_QUALITY_STD, size=n), 0.0, 1.0)
    
    df = pd.DataFrame({
        "supplier_id": [f"SUP-{i:04d}" for i in range(1, n + 1)],
        "quality_latent": np.round(quality, 4),
        "on_time_delivery_rate": np.round(np.clip(quality + rng.normal(0, 0.05, size=n), 0.0, 1.0), 4),
        "rating": np.round(np.clip(1.0 + 4.0 * quality + rng.normal(0, 0.3, size=n), 1.0, 5.0), 2),
        "dispute_count": rng.poisson(np.clip((1.0 - quality) * 8.0, 0.1, None)),
        "data_source": DATA_SOURCE_TAG
    })
    return df

# ==========================================
# 3. EXECUTION
# ==========================================
rng = np.random.default_rng(RANDOM_SEED)

price_df = generate_price_trends(rng)
price_df.sort_values(by="date", inplace=True)
price_df.to_csv("data/price_trends_sim.csv", index=False)
print(f"Generated price_trends_sim.csv with {len(price_df)} rows (Max Date: {price_df['date'].max()})")

orders_df = generate_landed_cost_orders(rng, price_df)
orders_df.to_csv("data/landed_cost_sim.csv", index=False)
print(f"Generated landed_cost_sim.csv with {len(orders_df)} rows (Max Date: {orders_df['order_timestamp'].max()})")

suppliers_df = generate_supplier_trust(rng)
suppliers_df.to_csv("data/supplier_trust_sim.csv", index=False)
print(f"Generated supplier_trust_sim.csv with {len(suppliers_df)} rows")
```
